```python
import os
import sys
from openai import OpenAI
from pydantic import BaseModel, Field

#  Hardcode your OpenAI API Key string right here for this session
os.environ["OPENAI_API_KEY"] = "<Enter your APIs>"
#  Simulate setting up your GitHub credential token for safety checks
os.environ["GITHUB_PERSONAL_ACCESS_TOKEN"] = "<Enter your APIs>"

# =====================================================================
# 1. EMULATING THE GITHUB MCP PROTOCOL SCHEMAS
# =====================================================================
# This class acts as the structural tool advertised by the GitHub MCP Server
class GitHubCreateIssueSchema(BaseModel):
    owner: str = Field(description="The GitHub profile username or organization handle.")
    repo: str = Field(description="The target repository name string.")
    title: str = Field(description="The summary header title of the issue.")
    body: str = Field(description="The complete, descriptive body text detailing the bug report.")

# Validate that our environment vars are safely present before calling the client
if not os.environ.get("OPENAI_API_KEY") or not os.environ.get("GITHUB_PERSONAL_ACCESS_TOKEN"):
    print(" ERROR: Missing required security credentials in environment variables.")
else:
    # Initialize our AI execution interface client
    client = OpenAI()
    print("Cell 1 Executed: GitHub MCP tool schemas and validation protocols ready.")
```

```python
# =====================================================================
# 2. CORE AGENT PIPELINE EXECUTION
# =====================================================================
def run_github_automation_pipeline():
    print("--- Starting GitHub MCP Simulation Session ---")
    
    # Simulating a typical chaotic human developer message in Slack or an IDE chat
    user_prompt = (
        "Hey assistant, I noticed that our login authentication endpoint is throwing a "
        "500 Internal Error whenever users pass empty email strings. "
        "Can you log an issue in my repository 'amritanshl/EYAI' "
        "titled 'Critical Authentication Bug: Null Email Fix' so the team can track it?"
    )
    
    print(f"Incoming Developer Query:\n'{user_prompt}'\n")
    print(" AI is reading the prompt and checking attached MCP manifest capabilities...")

    # The client passes the prompt + the expected Pydantic schema to enforce strict format
    completion = client.beta.chat.completions.parse(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a senior DevOps agent. Map human bug notifications into strict GitHub Issue schema payloads."},
            {"role": "user", "content": user_prompt}
        ],
        response_format=GitHubCreateIssueSchema
    )
    
    # Extracting our validated structural parameters
    parsed_mcp_call = completion.choices[0].message.parsed
    
    print("\n[SUCCESS] AI successfully synthesized the GitHub MCP Tool Request Block:")
    print("-----------------------------------------------------------------")
    print(f"🛠️ Target Tool Name : `github_create_issue`")
    print(f"📦 Repository Owner: '{parsed_mcp_call.owner}'")
    print(f"📂 Target Repo Name: '{parsed_mcp_call.repo}'")
    print(f"📌 Synthesized Title: '{parsed_mcp_call.title}'")
    print(f"📝 Extracted Body   : '{parsed_mcp_call.body}'")
    print("-----------------------------------------------------------------")
    
    # Emulating crossing the bridge to execute the actual task locally or over the cloud
    print("\n🚀 Next Local Sub-process Step:")
    print(f"Passing payload to ghcr.io/github/github-mcp-server with authorization token header.")
    
    tool_name = "github_create_issue"
    print(f"🧾 MCP Tool Used: `{tool_name}`")

    repo_dir = os.path.join(os.getcwd(), parsed_mcp_call.repo)
    os.makedirs(repo_dir, exist_ok=True)
    sample_path = os.path.join(repo_dir, "sample.txt")

    with open(sample_path, "w", encoding="utf-8") as f:
        f.write(f"Repository: {parsed_mcp_call.owner}/{parsed_mcp_call.repo}\n")
        f.write(f"Issue Title: {parsed_mcp_call.title}\n")
        f.write(f"Issue Body: {parsed_mcp_call.body}\n")
        f.write(f"MCP Tool Used: {tool_name}\n")

    print(f"✅ Created sample file at: {sample_path}")
print("✅ Cell 2 Executed: Pipeline execution function successfully built.")
```

```python
# Trigger the complete GitHub MCP agent loop pipeline execution block
run_github_automation_pipeline()
```
