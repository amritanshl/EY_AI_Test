```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

print("=== STARTING VECTORIZATION & EMBEDDING LAB ===")

# Step 1: Simulating dense semantic embedding vectors (3-dimensional for visualization)
# Let our dimensions represent: [Is_Fruit, Is_Electronic, Is_Royal]
dimensions_key = ["Fruit-ness", "Tech-ness", "Royal-ness"]

# Let's define semantic coordinates for four concepts
vector_apple_fruit = np.array([[0.95, 0.05, 0.00]]) # High fruit, low tech
vector_iphone      = np.array([[0.02, 0.98, 0.01]]) # Low fruit, high tech
vector_king        = np.array([[0.00, 0.10, 0.99]]) # Low fruit, high royalty
vector_banana      = np.array([[0.91, 0.01, 0.00]]) # High fruit, low tech (close to apple!)

print(f"Vector for 'Apple Fruit': {vector_apple_fruit[0]}")
print(f"Vector for 'Banana':      {vector_banana[0]}")
print(f"Vector for 'iPhone':      {vector_iphone[0]}\n")
```

```python
# Step 2: Calculate Semantic Similarity using Cosine Similarity Math
def check_similarity(name1, vec1, name2, vec2):
    # Calculates the cosine similarity score between 0 and 1
    similarity = cosine_similarity(vec1, vec2)[0][0]
    print(f"Semantic Similarity between [{name1}] and [{name2}]: {similarity:.4f}")

# Let's run the comparisons to trace the math logic
print("--- Running Similarity Check ---")
check_similarity("Apple Fruit", vector_apple_fruit, "Banana", vector_banana)
check_similarity("Apple Fruit", vector_apple_fruit, "iPhone", vector_iphone)
check_similarity("iPhone", vector_iphone, "King", vector_king)
```

```python
print("\n=== UNDER THE HOOD: MANUAL CALCULATION ===")
# Let's manually perform the dot product and norm calculation for Apple and Banana
A = vector_apple_fruit[0]
# Using the standard mathematical text variables
B = vector_banana[0]

dot_product = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)
manual_cosine = dot_product / (norm_A * norm_B)

print(f"Dot Product (A · B):       {dot_product:.4f}")
# Magnitude length of the vector strings
print(f"Magnitude ||A||:           {norm_A:.4f}")
print(f"Magnitude ||B||:           {norm_B:.4f}")
print(f"Calculated Cosine Match:   {manual_cosine:.4f}")
```
